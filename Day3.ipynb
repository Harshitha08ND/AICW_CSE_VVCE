```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    @classmethod   #you have to use cls in the class method
    def calculate_age(cls,name,birthyear):
        return cls(name,date.today().year - birthyear)
    def show(self):
        print(self.name +"'s age is " +str(self.age))
```

```python
age=student('Harshitha',20)
age.show()
```

```python
by=student.calculate_age('Harshitha',2006)
by.show()
```

<h3>Mini Project</h3>

```python
#mini project
class BankAccount:
    def __init__(self,name,balance=0):
        #instance atttributes
        self.name=name
        self.balance=balance
    #deposit money
    def  deposit(self,amount):
        if amount>0:
            self.balance +=amount
            print("deposit successfully!")
            print("deposited:",amount)
        else:
            print("invalid deposit amount")
    #withdraw money
    def withdraw(self,amount):
        if amount<=0:
            print('invalid withdraw amount!')
        elif amount > self.balance:
            print('insufficient balance!')
        else:
            self.balance -= amount
            print("withdrawal successful!")
            print("withdrawn:",amount)
    #check balance
    def check_balance(self):
        print("current balance:",self.balance)
    #display account details
    def display_details(self):
        print("account holder:",self.name)
        print("balance:",self.balance)
```

```python
acc1=BankAccount('Harshitha',1000)
```

```python
acc1.display_details()
```

```python
acc1.deposit(5000)
```

```python
acc1.check_balance()
```

```python
acc1.withdraw(3000)
```

```python
acc1.check_balance()
```

<h3>Inheritance,Polymorphism,Encapsulation</h3>

```python
#parent class
class Person:
    def person_info(self,name,age):
        print('Inside Person Class')
        print('Name:',name,'Age:',age)
#parent class 2
class Company:
    def company_info(self,company_name,location):
        print('Inside the company class')
        print('Name:',company_name,'Location:',location)
```

```python
#child class
class Employee(Person,Company):
    def Employee_info(self,salary,skill):
        print('Inside employee class-child class')
        print('Salary',salary,'Skills:',skill)
```

```python
#create an object of employee
emp2=Employee()
emp2.person_info('Harshitha',20)
emp2.company_info('Edunet foundation','Bengluru')
emp2.Employee_info(25000,'AIML Engineer')
```

```python
#hybrid inheritance
class Vehicle:
    def vehicle_info(self):
        print('Inside vehicle class')
class Car(Vehicle):
    def car_info(self):
        print('Inside car class')
class Truck(Vehicle):
    def truck_info(self):
        print('Inside truck class')
class Sportscar(Car,Vehicle):
    def sports_car_info(self):
        print('Inside sportscar class')
```

```python
#Super Class
class Company:
    def company_name(self):
        return 'Edunet Foundation'
class Employee(Company):
    def emp_info(self):
        #call the super function
        c_name=super().company_name() #calls the parent class
        print('RMS works at',c_name)
```

```python
emp3=Employee()
emp3.emp_info()
```

```python
#public member
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    #public instance method
    def show(self):
        #access public data members
        print('Name',self.name,'Salary',self.salary)
```

```python
emp=Employee('Harshitha',10000)
#access the public data members
print('Name',emp.name,'Salary',emp.salary)
emp.show()
```

```python
#Private member
class Employee:
    def __init__(self,name,salary):
        self.name=name #public data member
        self.__salary=salary #private data member
        
```

```python
#create an object
emp=Employee('harshitha',5000)
#access the private data member
print('Salary',emp.__salary) #even if you the attribute name you cannot access the private data member
```

```python
#protected member
class Company:
    def __init__(self):
        #protected member
        self._project="NLP"
#Child class
class Employee(Company):
    def __init__(self,name):
        self.name=name
        Company.__init__(self)
    def show(self):
        print("Employee name:",self.name)
        #Access the protected member
        print('working on project:',self._project)
```

```python
c=Employee("Harshitha")
c.show()
#direct access
print("project:",c._project)
```

```python
class India():
    def capital(self):
        print("New Delhi is the capital of India")
    def language(self):
        print("India has 22 languages")
    def type(self):
        print("India is a developing nation!")
class USA():
    def capital(self):
        print("Capital of USA is Washington DC")
    def language(self):
        print("The official language in US is English")
    def type(self):
        print("USA is a developed nation!")
```

```python
class Car:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("Drive!")
class Boat:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("Sail!")
class Plane:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("Fly!")
```

```python
#polymorphism with inheritance
class Bird:
    def intro(self):
        print('There are many types of birds!')
    def flight(self):
        print('most of the birds can fly but some cannot!')
class sparrow(Bird):
    def flight(self):
        print("Sparrow can fly!")
class ostrich(Bird):
    def flight(self):
        print("Ostrich cannot fly!")
```

```python
ob_bird=Bird()
ob_bird.intro()
ob_bird.flight()
```

```python
ob_sparrow=sparrow()
ob_sparrow.flight()
```

```python
ob_ost=ostrich()
ob_ost.flight()
```

```python
#overloading 
class complex:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def __add__(self,other):
        return self.a  +other.a,self.b + other.b
ob1=complex(1,2)
ob2=complex(2,3)
ob3=ob1+ob2
print(ob3)
```
