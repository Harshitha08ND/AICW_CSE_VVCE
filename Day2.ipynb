```python
#__init__ constructor -used to initialise
class Person:
    def __init__(self,name,age,school):
        self.name=name
        self.age=age
        self.school=school
    def myfun(self):
        print('Hello my name is',self.name)
        print('my age is',self.age)
        print('i went to',self.school)
```

```python
p1=Person('Madhu',38,'BVB')
p1.myfun()
```

```python
class Student:
    def __init__(self,usn,name,age):
        #instance variable 
        self.usn=usn
        self.name=name
        self.age=age
    #instance method access instance variable
    def show(self):
        print('USN',self.usn,'Name',self.name,'Age',self.age)
    #instance method to modify instance variable
    def update(self,usn,age):
        self.usn=usn
        self.age=age
        
```

```python
#create an object
stu=Student(109,'Harshitha',20)
```

```python
stu.show()
```

```python
stu.update(101,19)
```

```python
stu.show()
```
